```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel('data_pertemuan2.xlsx')
df
```

```python
df.info()
```

```python
df.describe().T
```

**Statistik Jenis Kelamin**

```python
# Ubah nilai sex, 0 menjadi Wanita, 1 menjadi Pria
df["sex"] = df["sex"].apply(lambda x: "Wanita" if x == 0 else "Pria")
df['sex']
```

```python
sex_count = df['sex'].value_counts()
sex_count
```

```python
def format(x):
    print(x)
    return '{:.1f}%\n{:.0f}'.format(x, sum(sex_count)*x/100)

plt.pie(sex_count, labels=['Pria','Wanita'], autopct=format)
plt.title("Persentase Jenis Kelamin")
plt.show()
```

**Statisik Usia**

```python
sns.histplot(df['age'], kde=True, color='cornflowerblue')
plt.title(f"Sebaran Nilai Usia")
```

```python
print(df['age'].describe())
print()

modus_age = df['age'].mode()
print("Modus Usia :\n", modus_age)
```

**Statistik Kolesterol**

```python
sns.histplot(df['chol'], kde=True, color='cornflowerblue')
plt.title(f"Sebaran Nilai Kolesterol Serum (chol) dalam mg/dl")
```

```python
print(df['chol'].describe())
print()

modus_chol = df['chol'].mode()
print("Modus Kolesterol :\n", modus_chol)
```

**Statistik Tekanan Darah Istirahat**

**Statistik Nyeri Dada**

**Statistik Target**

```python
target_count = df['target'].value_counts()
target_count
```

```python
def format(x):
    print(x)
    return '{:.1f}%\n{:.0f}'.format(x, sum(target_count)*x/100)

plt.pie(target_count, labels=['Sakit Jantung','Sehat'], autopct=format)
plt.title("Persentase Sakit Jantung")
plt.show()
```

**Distribusi jenis nyeri dada berdasarkan jenis kelamin**

```python
cp_per_sex = df.groupby(["sex", "cp"]).count()
cp_per_sex
print(cp_per_sex['target'])
```

```python
g = sns.catplot(
    data=cp_per_sex, kind="bar",
    x="sex", y=cp_per_sex['target'], hue="cp",
    palette=sns.color_palette(['green', 'blue', 'yellow', 'red']), height=5
)
for ax in g.axes.ravel():
    for c in ax.containers:
        ax.bar_label(c, label_type='edge')

plt.title("Distribusi jenis nyeri dada berdasarkan jenis kelamin")
plt.xlabel("Jenis Kelamin")
plt.ylabel("Count")
plt.show()
```
